```python
arr = [10, 20, 30, 40, 50]
total = 0
for val in arr:
  total += val
print("Sum =", total)
```

```python
arr = [12, 45, 23, 51, 19, 8]
max_val = arr[0]
for num in arr:
  if num > max_val:
    max_val = num

print("Maximum Element:", max_val)
```

```python
string = "Python"
reversed_str = ""
for ch in string:
  reversed_str = ch + reversed_str
print("Reversed:", reversed_str)
```

```python
arr = [1, 2, 2, 3, 1, 4, 2]
freq = {}
for num in arr:
  freq[num] = freq.get(num, 0) + 1
print(freq)
```

```python
arr = [4, 7, 1, 9, 3]
key = 9
found = False
for val in arr:
  if val == key:
    found = True
  break

print("Element Found!" if found else "Not Found")
```

```python
n = 5
fact = 1
for i in range(1, n+1):
  fact *= i
print("Factorial of", n, "is", fact)
```

```python
a, b, c = 10, 25, 15
if a > b:
  if a > c:
    print("Largest:", a)
  else:
    print("Largest:", c)
else:
  if b > c:
    print("Largest:", b)
  else:
    print("Largest:", c)
```

```python
for i in range(1, 4): # Outer loop for rows
  for j in range(1, 4): # Inner loop for columns
    print(j, end=" ")
print() # Move to next line after inner loop
```

```python
# Program 1: Right-Angled Triangle
for i in range(1, 6):
    for j in range(1, i + 1):
      print("*", end=" ")
print()
```

```python
# Program 2: Inverted Triangle

for i in range(5, 0, -1):
  for j in range(1, i + 1):
    print("*", end=" ")
print()
```

```python
# Program 3: Number Pattern
for i in range(1, 6):
    for j in range(1, i + 1):
      print(j, end=" ")
print()
```

```python
# Program 4: Pyramid Pattern
rows = 5
for i in range(1, rows + 1):
    print(" " * (rows - i), end="") # Spaces
print("* " * i)
```

```python
# Program 6: Alphabet Triangle
ch = 65 # ASCII for 'A'
for i in range(5):
  for j in range(i + 1):
    print(chr(ch), end=" ")
ch += 1
print()
```

```python
# Program 7: Floyd's Triangle
num = 1
for i in range(1, 6):
  for j in range(1, i + 1):
    print(num, end=" ")
num += 1
print()
```

```python
# Program 8: Diamond Pattern
rows = 5
# Upper part
for i in range(1, rows + 1):
  print(" " * (rows - i) + "* " * i)
# Lower part
for i in range(rows - 1, 0, -1):
  print(" " * (rows - i) + "* " * i)
```
